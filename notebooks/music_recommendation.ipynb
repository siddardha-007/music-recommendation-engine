```python
import pandas as pd

df = pd.read_csv("../data/raw/spotify_tracks.csv")

print(df.shape)
print(df.head())
print(df.info())
```

```python
print(df.isnull().sum())
print(df.duplicated().sum())
```

```python
audio_features = [
    "danceability",
    "energy",
    "loudness",
    "speechiness",
    "acousticness",
    "instrumentalness",
    "liveness",
    "valence",
    "tempo"
]

print(df[audio_features].describe())
```

```python
for col in audio_features:
    print(
        col,
        "min =", df[col].min(),
        "max =", df[col].max()
    )
```

```python
for col in audio_features:
    print(
        col,
        "min =", df[col].min(),
        "max =", df[col].max()
    )
```

```python
df = df.drop_duplicates(
    subset=["track_id"]
).reset_index(drop=True)
```

```python
df[audio_features].isnull().sum()
```

```python
df = df.dropna(
    subset=audio_features
).reset_index(drop=True)
```

```python
import matplotlib.pyplot as plt

df[audio_features].hist(
    figsize=(15, 12),
    bins=30
)

plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 8))

sns.heatmap(
    df[audio_features].corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Audio Feature Correlation")
plt.show()
```
