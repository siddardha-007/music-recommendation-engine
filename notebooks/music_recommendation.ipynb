```python
import pandas as pd

df = pd.read_csv("../data/raw/spotify_tracks.csv")

print(df.shape)
print(df.head())
print(df.info())
```

```python
print(df.isnull().sum())
print(df.duplicated().sum())
```

```python
audio_features = [
    "danceability",
    "energy",
    "loudness",
    "speechiness",
    "acousticness",
    "instrumentalness",
    "liveness",
    "valence",
    "tempo"
]

print(df[audio_features].describe())
```

```python
for col in audio_features:
    print(
        col,
        "min =", df[col].min(),
        "max =", df[col].max()
    )
```
